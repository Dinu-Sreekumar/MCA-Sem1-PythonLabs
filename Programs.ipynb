# 6. Working With tuples

```python
# Creating Empty list
t1=()
# Tuple with mixed datatypes
t2=(1,"a",2,"b",3,"c")
#Single element tuple
t3=("Great",)
# Accessing Elements (Indexing & Slicing)
my_tuple = ("apple", 42, 3.14, "banana", True)
print(f"First element: {my_tuple[0]}")
print(f"Last element: {my_tuple[-1]}")
print(f"Sliced elements (index 1 to 3): {my_tuple[1:4]}\n")
# Built-in Tuple Methods
numbers_tuple = (1, 2, 3, 2, 4, 2, 5)
print(f"Count of '2': {numbers_tuple.count(2)}")
print(f"Index of first occurrence of '4': {numbers_tuple.index(4)}\n")
# Tuple Unpacking
coordinates = (10.0, 20.5, 30.1)
x, y, z = coordinates
print(f"X coordinate: {x}")
print(f"Y coordinate: {y}")
print(f"Z coordinate: {z}\n")
# Concatenation and Repetition
tuple_a = (1, 2, 3)
tuple_b = (4, 5, 6)
combined = tuple_a + tuple_b
repeated = tuple_a * 2
print(f"Concatenation: {combined}")
print(f"Repetition: {repeated}")
```

# 7.Working with dictionaries

```python
# Creating a dictionary
student = {"name": "Alice", "age": 20}
print("Created Dictionary:", student)

# Inserting into a dictionary (adding new key-value pairs)
student["grade"] = "A"
student["city"] = "New York"
print("After Insertion:", student)

# Taking specific values out
print("Name:", student["name"])
print("Age:", student.get("age"))

# Extracting keys, values, and key-value pairs separately
print("Keys:", list(student.keys()))
print("Values:", list(student.values()))
print("Key-Value Pairs:", list(student.items()))
```

# 8. Find the largest of 3 numbers

```python
num1=int(input("Enter the first number:"))
num2=int(input("Enter the Second number:"))
num3=int(input("Enter the Third number:"))
if num1==num2 and num2==num3:
    print("All the 3 numbers are equal!")
elif num1>=num2 and num1>num3:
    print(f"{num1} is the greatest")
elif num2>=num1 and num2>num3:
    print(f"{num2} is the greatest")
elif num3>=num1 and num3>num2:
    print(f"{num3} is the greatest")
else:
    print("Cannot determine the largest!")
```

# 9.Write a python program to convert temprature to and from Celsius to Fahrenheit

```python
choice=int(input("Enter whether to convert:\n 1.Celsius to Fahrenheit \n 2.Fahrenheit to Celsius \n Choice:"))
if choice==1:
    celsius=float(input("Enter the Celsius:"))
    Fahrenheit=(celsius*9/5)+32
    print(f"The Fahrenheit is:{Fahrenheit}")
elif choice==2:
    Fahrenheit=float(input("Enter the Fahrenheit:"))
    celsius=(Fahrenheit-32)* 5/9
    print(f"The Celsius is:{celsius}")
else:
    print("You entered wrong choice! Please run the program again!")
```

# 11.Write a python program to print prime numbers less than 20

```python
for num in range(2,20):
    isPrime=True
    for i in range(2,num):
        if num%i==0:
            isPrime=False
            break
    if isPrime==True:
        print(num)
```

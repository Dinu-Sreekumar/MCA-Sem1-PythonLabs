# 8. Find the largest of 3 numbers

```python
num1=int(input("Enter the first number:"))
num2=int(input("Enter the Second number:"))
num3=int(input("Enter the Third number:"))
if num1==num2 and num2==num3:
    print("All the 3 numbers are equal!")
elif num1>=num2 and num1>num3:
    print(f"{num1} is the greatest")
elif num2>=num1 and num2>num3:
    print(f"{num2} is the greatest")
elif num3>=num1 and num3>num2:
    print(f"{num3} is the greatest")
else:
    print("Cannot determine the largest!")
```

# 9.Write a python program to convert temprature to and from Celsius to Fahrenheit

```python
choice=int(input("Enter whether to convert:\n 1.Celsius to Fahrenheit \n 2.Fahrenheit to Celsius \n Choice:"))
if choice==1:
    celsius=float(input("Enter the Celsius:"))
    Fahrenheit=(celsius*9/5)+32
    print(f"The Fahrenheit is:{Fahrenheit}")
elif choice==2:
    Fahrenheit=float(input("Enter the Fahrenheit:"))
    celsius=(Fahrenheit-32)* 5/9
    print(f"The Celsius is:{celsius}")
else:
    print("You entered wrong choice! Please run the program again!")
```

# 11.Write a python program to print prime numbers less than 20

```python
for num in range(2,20):
    isPrime=True
    for i in range(2,num):
        if num%i==0:
            isPrime=False
            break
    if isPrime==True:
        print(num)
```
